## Step 2: Data Cleaning
Handle missing values:
* Age → median
* Embarked → mode
* Cabin → drop
* remove duplicates if any

```python
# check if there any outlier
train.duplicated().sum()
```

* it returns boolean if there is duplicated it returns True and if no returns False
* `.sum()` return 0 which indicates there is no dupplicated records. SO, **Each passenger entry is unique**

```python
# i defined catagorical and numerical features for simplicity to  see which feature is catagorical for reference

catagorical=train.select_dtypes(include="object").columns
numerical=train.select_dtypes(exclude="object").columns
```

```python
numerical
```

`Age` is numerical data type and have 177 missing values fill with mean or median so first let's check the distribution is there any outlier or not because it's matter alot.

**Because of the contraints use pandas only i can't plot the distribution. let me use IQR method to detect outlier**

```python
#outlier detection usig IQR 
# define lower and upper bound to check the value is below the lower bound or above the upper bound

q1=train["Age"].quantile(0.25)
q3=train["Age"].quantile(0.75)
IQR=q3-q1
lower_bound=q1 - 1.5*IQR
upper_bound=q3 + 1.5*IQR
outlier=train[(train["Age"] < lower_bound) | (train["Age"] > upper_bound)]
outlier["Age"]
```

`Age` columns have outliers filling with mean leads to `False` implication to the Age because mean is sensetive to outliers i should use median. 

```python
# Age fill with pandas i can't use scikit to impute

train["Age"] = train["Age"].fillna(train["Age"].median())
train["Age"].isnull().sum()
```

Thanks for Pandas i easily impute the `Age` column. No missing values

```python
catagorical
```

* `Cabin` and `Embarked` columns have object data type.
* **decision** impute `Cabin`  with mode and drop the `Cabin` columns

```python
# Embarked catagorical column with 2 miss we can fill with mode

train["Embarked"] = train["Embarked"].fillna(train["Embarked"].mode)
train["Embarked"].isnull().sum()
```

**all missings are handled with most frequent**

```python
# Drop Cabin column
train.drop("Cabin",axis=1,inplace=True)
```

## Step 3: Data Analysis
Using `groupby()`:
* Survival rate by gender
* Survival rate by class
* Average age per class
* Survival rate by age group


```python
# Survival rate by gender
train.groupby("Sex")["Survived"].mean()
```

**observation:**
* Survival rate of Female is 74%
* survival rate of male 19%
  
**Insights:**
- **Women had a much higher survival rate** compared to men.
- This reflects the "women and children first" Hypothesis.

**Decision:** Gender should be considered as a strong predictor in survival models.

```python
# Survival rate by class
train.groupby("Pclass")["Survived"].mean()
```

**observation**
Survival chances decreased with lower class.
* 1st class 63% the highest survival rate
* 2nd class 47% moderate survival rate.
* 3rd class 24% the lowest survival rate.
  
**Insights** 
* This highlights the **strong influence of socio-economic status** on survival chances. Higher‑class passengers had better access to lifeboats and crew assistance. 
 
**Decision:** Passenger class strongly influenced survival.so, `Pclass` is an important feature for predictive modeling.

```python
# Average Age per Class
train.groupby("Pclass")["Age"].mean()
```

**Observation:** Age distribution varies across classes.

**Insight:** Higher classes may have slightly older passengers.

**Decision:** Age combined with class could refine survival predictions.

```python
# Survival Rate by Age Group
# first le's create Age group using `.cut()`

train["AgeGroup"] = pd.cut(train["Age"], bins=[0, 12, 18, 35, 60, 100],
                           labels=["Child", "Teen", "Adult", "Middle-aged", "Senior"])
# 
train.groupby("AgeGroup",observed=True)["Survived"].mean()       # in the future default observed= False will deprecated
```

**Observation:**
- Children had the highest survival rate ~58% among age groups.  
- Seniors had the lowest survival rate.  
- Teens, adults, and middle-aged passengers had moderate survival rates.  

**Insight:**
- **Younger passengers were prioritized** during evacuation, consistent with the "women and children first" policy.  
- Older passengers faced lower survival chances, possibly due to mobility challenges or evacuation priorities.  

**Decision:**
- Age grouping is a valuable feature for predictive modeling.  
- Combining  with others features will likely improve survival prediction accuracy in Modeling.
